```python
import pandas as pd
import numpy as np
```

# •	Extraction

```python
# Load datasets from Excel sheets
xls= pd.ExcelFile("CRM_Analytics_Datasets.xlsx") 
xls.sheet_names
```

```python
contacts_df = pd.read_excel(xls, sheet_name="contacts")
contacts_df.head()
```

```python
deals_df = pd.read_excel(xls, sheet_name="deals")
deals_df.head()
```

```python
activities_df = pd.read_excel(xls, sheet_name="activities")
activities_df.head()
```

```python
# Building Leads sheet
contact_from_deals = deals_df['contact_id'].unique()
leads_df = contacts_df[contacts_df['contact_id'].isin(contact_from_deals)]
leads_df.head()
```

# •	Cleaning & Standardization


```python
# Handling missing vallues
contacts_df.isnull().sum()
```

```python
# Handling missing email
contacts_df = contacts_df.dropna(subset=["email"])
```

```python
deals_df.isnull().sum()
```

```python
activities_df.isnull().sum()
```

```python
leads_df.isnull().sum()
```

```python
# Convert closed_date to datetime
deals_df["closed_date"] = pd.to_datetime(deals_df["closed_date"], errors="coerce")
deals_df["closed_date"]
```

```python
# Convert created_date to datetime from contact sheet
contacts_df["created_date"] = pd.to_datetime(contacts_df["created_date"], errors="coerce")
contacts_df["created_date"]
```

```python
# Convert created_date to datetime from deal sheet
deals_df["created_date"] = pd.to_datetime(deals_df["created_date"], errors="coerce")
deals_df["created_date"]
```

```python
# Remove duplicate deals
contacts_df = contacts_df.drop_duplicates(subset=["contact_id"])
deals_df = deals_df.drop_duplicates(subset=["deal_id"])
activities_df = activities_df.drop_duplicates(subset=["activity_id"])
```

```python
# Convert deal_amount to numeric
deals_df["deal_amount"] = pd.to_numeric(deals_df["deal_amount"], errors="coerce")
```

```python
#Handling deal amount
deals_df["deal_amount"] = deals_df["deal_amount"].fillna(0)
```

```python
# Ensure IDs are integers
deals_df["deal_id"] = deals_df["deal_id"].astype(int)
contacts_df["contact_id"] = contacts_df["contact_id"].astype(int)
activities_df["activity_id"] = activities_df["activity_id"].astype(int)
```

```python
# Rename created_date columns according to their sheet names
leads_df.rename(columns={"created_date": "lead_created_date"}, inplace=True)
contacts_df.rename(columns={"created_date": "contact_created_date"}, inplace=True)
deals_df.rename(columns={"created_date": "deal_created_date"}, inplace=True)
```

```python
#Rename sales_rep column name according to their sheet name
deals_df.rename(columns={"sales_rep": "deals_sales_rep"},  inplace=True)
activities_df.rename(columns={"sales_rep": "activities_sales_rep"},  inplace=True)
```

# •	Merge HubSpot and Salesforce-style datasets into a unified structure

```python
# Merge leads with contacts
crm_df = leads_df.merge(
    contacts_df,
    on="contact_id",
    how="left"
)
```

```python
# Merge with deals
crm_df = crm_df.merge(
    deals_df,
    on="contact_id",
    how="left"
)
```

```python
# Merge with activities
crm_df = crm_df.merge(
    activities_df,
    on="contact_id",
    how="left"
)
```

```python
# Final Unified CRM DataFrame
crm_df.head()
```

```python
# Drop duplicate columns
crm_df = crm_df.drop(columns=[
    "first_name_y",
    "last_name_y",
    "email_y",
    "lead_source_y",
    "company_name_y"
])
```

```python
# Rename _x columns to clean names
crm_df = crm_df.rename(columns={
    "first_name_x": "first_name",
    "last_name_x": "last_name",
    "email_x": "email",
    "company_name_x": "company_name",
    "lead_source_x": "lead_source"
})
```

```python
crm_df.head()
```

```python
print(crm_df.info())
```

# Computed columns

```python
# Lead Response Time in hours
crm_df['lead_response_time_hours'] = (
    crm_df['response_time'] - crm_df['contact_time']
).dt.total_seconds() / 3600
crm_df['lead_response_time_hours']
```

```python
#Deal Conversion Rate per Sales Representative
crm_df.columns = crm_df.columns.str.strip().str.lower()
```

```python
closed_won_flag= crm_df['deal_stage'].eq('Closed Won').astype(int)
closed_won_flag
```

```python
closed_won_total= closed_won_flag.groupby(crm_df["deals_sales_rep"]).transform("sum")
total_deals= crm_df.groupby("deals_sales_rep")["contact_id"].transform("count")
       
crm_df["conversion_rate"] = closed_won_total/total_deals
```

```python
crm_df.head(50)
```

```python
# Customer Lifetime Value (CLV)
crm_df['clv']=(crm_df.where(crm_df['deal_stage']=='Closed Won')
               .groupby('contact_id')['deal_amount'].transform('sum'))

crm_df['clv'].head(20)
```

```python
crm_df['clv']=crm_df['clv'].fillna(0)
```
